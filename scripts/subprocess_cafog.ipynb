```python

import subprocess
import os

# Base folder
base_folder = r"C:\Users\b1095820\OneDrive - Universität Salzburg\Documents\DGTX\project_doe_2026\analysis\cafog"


# Get only directories
folder_names = [
    f for f in os.listdir(base_folder)
    if os.path.isdir(os.path.join(base_folder, f))
]

for folder_name in folder_names:
    folder_path = os.path.join(base_folder, folder_name)

    print(f"\n--- Analyzing: {folder_name} ---")

    result = subprocess.run([
        "python", "cafog.py",
        "-f", os.path.join(folder_path, "glycosylation.csv"),
        "-g", os.path.join(folder_path, "glycation.csv"),
        "-l", os.path.join(folder_path, "glycan_library.csv"),
    ], capture_output=True, text=True)

    print("RETURN CODE:", result.returncode)

    if result.stdout:
        print("STDOUT:\n", result.stdout[:500])  # limit long output

    if result.stderr:
        print("STDERR:\n", result.stderr[:500])

    # Save output if successful
    if result.returncode == 0:
        output_file = os.path.join(folder_path, "results.csv")
        with open(output_file, "w") as f:
            f.write(result.stdout)
        print(f"✔ Saved results to {output_file}")
    else:
        print(f"❌ Error in {folder_name}, skipping file write")
```
